```python
import numpy as np # linear algebra
import pandas as pd #
data = pd.read_csv('crimedata.csv')
data
```

```python
# 删除缺失值较多的列（例如缺失值超过80%的列）
threshold = len(data) * 0.8
data = data.dropna(thresh=threshold, axis=1)

# 填充剩余的缺失值（用均值填充数值列，用众数填充类别列）
for col in data.columns:
    if data[col].dtype == "object":
        data[col].fillna(data[col].mode()[0], inplace=True)
    else:
        data[col].fillna(data[col].mean(), inplace=True)
        
```

```python
# 频率编码
community_freq = data["communityName"].value_counts(normalize=True)
data["communityName_freq"] = data["communityName"].map(community_freq)
data = data.drop("communityName", axis=1)
```

```python
from category_encoders import TargetEncoder

# 使用目标编码
encoder = TargetEncoder()
data["state_encoded"] = encoder.fit_transform(data["state"], data["ViolentCrimesPerPop"])
data = data.drop("state", axis=1)  # 删除原始的 state 列
```
