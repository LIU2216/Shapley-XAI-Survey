```python
import matplotlib.pyplot as plt
import numpy as np

# Feature labels and their corresponding Shapley values
labels = ['CreditScore', 'Geography', 'Gender', 'Age', 'Tenure', 
          'Balance', 'NumOfProducts', 'HasCrCard', 'IsActiveMember', 'EstimatedSalary']
values = np.array([-289.61423, 294.04172, 382.8509, 109.67176, -150.30894,
                   -697.87604, 659.55225, 172.93318, 398.25836, 5.651661], dtype=float)

# 排序，从大到小
sorted_indices = np.argsort(values)[::1]  # 获取从大到小的索引
sorted_labels = [labels[i] for i in sorted_indices]
sorted_values = values[sorted_indices]

# 创建一个特定大小的图形
plt.figure(figsize=(4, 4), facecolor='white')

# 绘制水平条形图
plt.barh(sorted_labels, sorted_values, color='orange')

# 添加坐标轴标签
plt.xlabel('SHAP Value')
plt.ylabel('Features')

# 调整布局以适应大小
plt.tight_layout()

# 保存高分辨率图像
plt.savefig('fastshap_bank_lgb.eps', format='eps', dpi=300, bbox_inches='tight')

# 显示图表
plt.show()
```
